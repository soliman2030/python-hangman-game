## Description
> ##### The Hangman program randomly selects a secret word from a list of secret words.
> ##### The Game: Here, a random word (a fruit name) is picked up from our collection and the player gets limited chances to guess the fruit name correctly.
> ##### When a letter in that word is guessed correctly, that letter position in the word is made visible in the output.
> ##### In this way, all letters of the word are to be guessed before all the chances are over.
> ##### For convenience, the number of chances equals the word length + 2.
> ##### For example, word to be guessed is mango, then user gets 5 + 2 = 7 chances, as mango is a five letter word.
> ##### Note : Total number of Guesses either right or wrong equals word length + 2.
> ##### Make game repeats until player ends the game.
> ##### Bonus : Create a function that takes 3 parameters `Easy`, `Medium` and `Hard` where `Medium` has more difficult words to be guessed than `Easy` and so on.

```python
import random


def get_word_by_difficulty(easy_words, medium_words, hard_words):
  """Bonus: Prompts the user to select a difficulty level and returns a random word

    from the corresponding list.
    """
  while True:
    choice = input("\nChoose difficulty (Easy, Medium, Hard): ").strip().lower()
    if choice == "easy":
      return random.choice(easy_words)
    elif choice == "medium":
      return random.choice(medium_words)
    elif choice == "hard":
      return random.choice(hard_words)
    else:
      print("Invalid choice. Please type Easy, Medium, or Hard.")


def play_hangman():
  # Word lists categorized by difficulty
  easy_fruits = ["apple", "mango", "grape", "peach", "lemon", "melon"]
  medium_fruits = [
      "banana",
      "orange",
      "papaya",
      "cherry",
      "apricot",
      "coconut",
  ]
  hard_fruits = [
      "pomegranate",
      "strawberry",
      "dragonfruit",
      "passionfruit",
      "persimmon",
      "blackberry",
  ]

  print("=== WELCOME TO FRUIT HANGMAN ===")

  # Outer loop to repeat the game until the player ends it
  while True:
    secret_word = get_word_by_difficulty(
        easy_fruits, medium_fruits, hard_fruits
    )
    chances = len(secret_word) + 2
    guessed_letters = []

    print(
        f"\nI have selected a fruit with {len(secret_word)} letters. You have"
        f" {chances} total guesses!"
    )

    # Inner loop for each round of guessing
    while chances > 0:
      # Build the current display string (e.g., "m _ n g _")
      display_word = [
          letter if letter in guessed_letters else "_" for letter in secret_word
      ]
      print("\nWord:", " ".join(display_word))
      print(f"Remaining chances: {chances}")
      if guessed_letters:
        print(f"Guessed so far: {', '.join(guessed_letters)}")

      # Check if all letters have been guessed
      if "_" not in display_word:
        print(f"\nCongratulations! You guessed the fruit: {secret_word.upper()}")
        break

      # Get player input
      guess = input("Guess a single letter: ").strip().lower()

      # Validate input (don't penalize invalid typos)
      if len(guess) != 1 or not guess.isalpha():
        print("Please enter a single valid alphabetical letter.")
        continue

      if guess in guessed_letters:
        print(f"You already guessed '{guess}'. Try a different letter.")
        continue

      # Record the valid guess and deduct 1 chance (as per rule: total right or wrong = len + 2)
      guessed_letters.append(guess)
      chances -= 1

      if guess in secret_word:
        print(f"Good job! '{guess}' is in the word.")
      else:
        print(f"Sorry, '{guess}' is not in the word.")

    # Check if player ran out of chances without finishing the word
    if chances == 0:
      final_check = [
          letter if letter in guessed_letters else "_" for letter in secret_word
      ]
      if "_" not in final_check:
        print(
            f"\nPhew! On your very last chance, you guessed: {secret_word.upper()}"
        )
      else:
        print(f"\nOut of chances! The secret fruit was: {secret_word.upper()}")

    # Ask to play again
    play_again = (
        input("\nDo you want to play again? (yes/no): ").strip().lower()
    )
    if play_again not in ["yes", "y"]:
      print("Thanks for playing Hangman! Goodbye.")
      break


# Run the game
play_hangman()
```
